```python
import importlib.util
import json
from pathlib import Path

spec = importlib.util.spec_from_file_location("quality_profile", "profile.py")
quality = importlib.util.module_from_spec(spec)
spec.loader.exec_module(quality)
snapshot = json.loads(Path("/tmp/iowa-quality-review/snapshot.json").read_text())
result = quality.profile(snapshot)
assert result["manifest_stable"]
result["stores"], result["generation"]
```
