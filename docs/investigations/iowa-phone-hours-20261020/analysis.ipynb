# Iowa phone and hours investigation

Read-only canonical branch snapshot, October 20 2026. Analysis includes missingness by segment, evidence selection, weekly day coverage, crawl URL gaps and strict retained source joins. See FINDINGS.md and sources.json for interpretation; candidates are not publication decisions.

```python
import importlib.util
import json
import sys
from pathlib import Path

repo = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "bord_food_api").is_dir())
sys.path.insert(0, str(repo))
h = repo / "docs/investigations/iowa-phone-hours-20261020"
spec = importlib.util.spec_from_file_location("iowa_phone_analysis", h / "analyze.py")
module = importlib.util.module_from_spec(spec)
spec.loader.exec_module(module)
snapshot = json.loads((h / "snapshot.json").read_text())
# Set this to the directory holding the three retained files identified by input-receipt.json.
inputs = Path("/tmp")


def records(name):
    return [
        json.loads(l)
        for l in (inputs / ("iowa-phone-hours-" + name + ".jsonl")).read_text().splitlines()
        if l.strip()
    ]


result = module.analyze(snapshot, records("candidates"), records("places"))
print(
    json.dumps(
        {
            k: result[k]
            for k in [
                "stores",
                "counts",
                "missing",
                "missing_hours_conflict_shapes",
                "missing_field_campaign_state",
            ]
        },
        indent=2,
    )
)
```
